```python
# 1. Get the age and convert to int
age = int(input("Enter the age: "))

# 2. Logic to set the STARTING price
if age < 12:
    price = 5
elif age >= 65:
    price = 7
else:
    price = 10

# 3. Ask about the discount
discount = input("Do you have a discount code? (yes/no): ")

# 4. Logic to CHANGE the price if they have a code
if discount == "yes":
    price = price - 2  # This takes the old price and makes it 2 dollars cheaper!

# 5. Show the result
print("Your final price is: $", price)
```

```python
# The Challenge: The Smart ATM
# You have $100 in your bank account. You want to withdraw some money.

# Variable: Create a variable called balance and set it to 100.

# Input: Ask the user: "How much money do you want to withdraw?"

# Hint: Remember to cast this to an integer!

# The Logic Check: * If the amount they want is less than or equal to their balance, they are allowed to take it.
# * Subtract the amount from the balance.
# * Print "Success! Your new balance is: [the new balance]".

# Else (if they ask for more than $100):

# Print "Error: You don't have enough money!"


balance = 100
user = int(input("How much money do you want to withdraw?"))
if user <= balance:
  print(f"Success! Your new balance is:{balance-user}")
else:
  print("Error: You don't have enough money!")

```

```python
# Level Up Challenge: The "Password & Withdrawal" Logic
# Let's combine everything you've learned: Strings, Integers, Type Casting, and Nested Ifs.

# The Goal: You can only withdraw money if your Password is correct AND you have Enough Money.

# Variables: balance = 100 and secret_pin = "1234"

# Step 1: Ask the user: "Enter your PIN: "

# Logic A: If the PIN is correct:

# Ask "How much to withdraw?"

# Logic B (Inside Logic A): If they have enough money, subtract it and print the new balance.

# If they don't have enough, print "Insufficient funds."

# Logic C: If the PIN was wrong at the start, just print "Incorrect PIN. Access Denied."


balance = 100
secret_pin = "1234"
user2 = int(input("How much to withdraw? "))
user1 = input("Enter your PIN: ")
if user1 == secret_pin:
  print("")
if user1 != secret_pin:
  print("Incorrect PIN. Access Denied")

elif user2 < balance:
    print(f"The New Balance is: {balance-user2}")
else:
    print("Insufficant funds")

#######################################
#                 OR
#######################################

# another option
balance = 100
secret_pin = "1234"

# 1. Ask for PIN FIRST
user_pin = input("Enter your PIN: ")

if user_pin == secret_pin:
    print("PIN Correct!")

    # 2. Ask for money ONLY IF the PIN was right
    withdraw = int(input("How much to withdraw? "))

    # 3. Check the balance (Inside the PIN block!)
    if withdraw <= balance:
        print(f"Success! New balance: {balance - withdraw}")
    else:
        print("Insufficient funds.")

else:
    # This runs if the FIRST 'if' (the PIN) was wrong
    print("Incorrect PIN. Access Denied.")


```

```python
# Challenge 2: The Username Formatter
# You want to create a full name from two separate inputs.

# Ask for first_name.

# Ask for last_name.

# Logic: Create a new variable called full_name by adding them together with a space in between.

# The Check: If the full_name is longer than 20 characters, print "Name is too long."

# Hint: Use len(full_name) to get the length.

first_name = input("Enter First Name ")
last_name = input("Enter last Name ")
full_name = first_name +" " + last_name
if len(full_name) > 20:
  print("Name is too long.")
else:
  print(f"The full name is {first_name} {last_name}")
```
